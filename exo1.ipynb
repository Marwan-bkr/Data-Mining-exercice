```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import where
import pandas as pd
```

```python
# 1.
print("1.")
xlsx = pd.ExcelFile('donnees_brutes_achats.xlsx')
df = pd.read_excel(xlsx, "Sheet1") 
print("")

# 2.
print("2.")
print("Infos :")
print(df.info())
print("")
print("Pourcentage de valeurs manquantes (%):")
print((df.isna().mean()*100).round(1), "\n")
print("Doublon(s) : ", len(df[df.duplicated()]), "/", len(df))
print("Number of unique TransactionID :", df['TransactionID'].nunique())
print("")

# 3.
print("3.")
df = df.drop_duplicates(subset=['TransactionID'])
print("Doublons TransactionID supprimés.")
print("")

# 4.
print("4.")
def clean(x):
    if isinstance(x, str):
        return x.strip()
    return x
df = df.apply(lambda col: col.map(clean))
print("trim all values")

df = df[df["Quantité"] > 0].copy()
df = df[df["Quantité"].notna()].copy()
print("Quantité <= 0 or NaN dropped")

df['Produit'] = (df['Produit'].str.lower().str.replace('é','e').str.replace('è','e').str.replace('ê','e').str.replace('à','a').str.replace('â','a').str.replace('î','i').str.replace('ï','i').str.replace('ô','o').str.replace('ö','o').str.replace('û','u').str.replace('ü','u'))
mapping_produits = {
    'pates':'pâtes',
    'the':'thé',
    'bananes':'banane',
    'coca-cola':'coca cola',
    'eau  minerale':'eau minerale',
    'oeufs':'oeuf',
    'cafe':'café',
    'tomato':'tomate',
    'pommes':'pomme',
    'yaourts':'yaourt',
    'riz':'riz',
    'frommage':'fromage',
    'huile olive':'huile d\'olive',
    'corn flakes':'corn-flakes'
}
df = df[df['Produit'] != '—'].copy()
df['Produit'] = df['Produit'].str.strip().str.lower().replace(mapping_produits)
print("Values of Produits uniformized")
print("")

# 5.
print("5.")
df['Catégorie'] = df['Catégorie'].str.strip().str.lower().replace({
    'epicerie':'épicerie',
    'boisson':'boissons',
    'fruits-legumes':'fruits & légumes',
    'fruits et legumes':'fruits & légumes',
    'fruits/legumes':'fruits & légumes',
    'fruits/légumes':'fruits & légumes',
    'laitage':'laitage',
    'cremerie':'crèmerie',
    'oeufs':'œufs & ovoproduits'})
print("Values of Catégorie uniformized")
print("")

# 6.
print("6.")
def cleanDate(x):
    if ('/' in x):
        return pd.to_datetime(x, dayfirst=True, errors='coerce')
    elif ('-' in x):
        return pd.to_datetime(x, errors='coerce')
    return x
df["Date"] = df["Date"].map(cleanDate)
print("Date uniformized")
print("")

# 7.
print("7.")
print("Quantité already cleaned in step 4.")
print("")

# 8.
print("8.")
print("Description of all columns : ")
for col in df.columns:
    print(f"{col}:")
    print(df[col].describe())
    print("")
print("Quantité and Prix outliers replaced by Quantité and Prix median without outliers: ")
df["Quantité"] = df["Quantité"].replace(df["Quantité"].max(), df["Quantité"].where(df["Quantité"] < 100).dropna().median())
df["Prix"] = df["Prix"].replace(df["Prix"].max(), df["Prix"].where(df["Prix"] < 100).dropna().median())
print("")

# 9.
print("9.")
print(df["Notes"].unique())
df = df.drop(columns=['Notes'], axis=1)
print('Notes column dropped')

# 10.
print("10.")
print(df.info())
df.to_csv("donnees_achats_propre.csv", index=False)
print("Cleaned data saved into donnees_achats_propre.csv")
```
